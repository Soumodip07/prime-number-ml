```python
import numpy as np
import pandas as pd

from pathlib import Path

BASE_DIR = Path("E:/Python/prime_number_project/datasets")
```

```python
def generate_primes(n):
    sieve = [True] * (n + 1)
    sieve[0] = sieve[1] = False

    for i in range(2, int(n**0.5) + 1):
        if sieve[i]:
            for j in range(i * i, n + 1, i):
                sieve[j] = False

    return [i for i in range(n + 1) if sieve[i]]


def create_dataset(start, end, save_path, global_max):

    max_bits = len(bin(global_max)[2:])

    primes = generate_primes(int(np.sqrt(global_max)) + 1)

    data = []

    for n in range(start, end + 1):

        binary = format(n, f"0{max_bits}b")

        is_prime = 1
        for p in primes:
            if p * p > n:
                break
            if n % p == 0:
                is_prime = 0
                break

        mod2  = (n % 2) / 1 
        mod3  = (n % 3) / 2
        mod5  = (n % 5) / 4
        mod7  = (n % 7) / 6
        mod11 = (n % 11) / 10
        mod13 = (n % 13) / 12
        mod17 = (n % 17) / 16
        mod19 = (n % 19) / 18
        mod23 = (n % 23) / 22
        mod29 = (n % 29) / 28
        mod31 = (n % 31) / 30

        data.append([
            n, binary,
            mod2, mod3, mod5, mod7, mod11,
            mod13, mod17, mod19, mod23, mod29, mod31,
            is_prime
        ])

    df = pd.DataFrame(data,
        columns=[
            "number", "binary",
            "mod2","mod3","mod5","mod7","mod11",
            "mod13","mod17","mod19","mod23","mod29","mod31",
            "is_prime"
        ]
    )

    df["is_prime"] = df["is_prime"].astype(int)

    df.to_csv(save_path, index=False)

    print(f"\n Saved dataset: {save_path}")
    print(df.head())
```
